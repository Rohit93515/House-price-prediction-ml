```python
# Remove ID because it is just an identifier
X = df.drop(columns=["Property_ID", "Price_INR_Lakhs"])

# Target: house price in lakhs
y = df["Price_INR_Lakhs"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)

print("Categorical columns:")
print(X.select_dtypes(include="object").columns.tolist())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training rows:", X_train.shape[0])
print("Testing rows:", X_test.shape[0])
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Identify categorical and numerical features
categorical_cols = X.select_dtypes(include="object").columns.tolist()
numerical_cols = X.select_dtypes(include="number").columns.tolist()

# Convert text categories into numerical columns
preprocessor = ColumnTransformer(
    transformers=[
        ("categorical", OneHotEncoder(handle_unknown="ignore"), categorical_cols),
        ("numerical", "passthrough", numerical_cols)
    ]
)

# Create complete ML pipeline
model = Pipeline([
    ("preprocessor", preprocessor),
    ("regressor", LinearRegression())
])

# Train the model
model.fit(X_train, y_train)

# Predict prices on unseen test data
y_pred = model.predict(X_test)

# Evaluate model
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("MAE:", round(mae, 2), "lakhs")
print("RMSE:", round(rmse, 2), "lakhs")
print("R2 Score:", round(r2, 4))
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Preprocessing
preprocessor_rf = ColumnTransformer([
    ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_cols),
    ("num", "passthrough", numerical_cols)
])

# Random Forest model
rf_model = Pipeline([
    ("preprocessor", preprocessor_rf),
    ("regressor", RandomForestRegressor(
        n_estimators=100,
        min_samples_leaf=2,
        random_state=42,
        n_jobs=-1
    ))
])

# Train
rf_model.fit(X_train, y_train)

# Predict
rf_pred = rf_model.predict(X_test)

# Evaluate
rf_mae = mean_absolute_error(y_test, rf_pred)
rf_rmse = np.sqrt(mean_squared_error(y_test, rf_pred))
rf_r2 = r2_score(y_test, rf_pred)

print("Random Forest Results")
print("MAE:", round(rf_mae, 2), "lakhs")
print("RMSE:", round(rf_rmse, 2), "lakhs")
print("R2 Score:", round(rf_r2, 4))
```

```python
results = pd.DataFrame({
    "Model": ["Linear Regression", "Random Forest"],
    "MAE": [mae, rf_mae],
    "RMSE": [rmse, rf_rmse],
    "R2 Score": [r2, rf_r2]
})

display(results.round(3))
```

```python
# Get the transformed feature names after One-Hot Encoding
feature_names = (
    rf_model.named_steps["preprocessor"]
    .get_feature_names_out()
)

# Get feature importance from Random Forest
importances = (
    rf_model.named_steps["regressor"]
    .feature_importances_
)

# Create a DataFrame
importance_df = pd.DataFrame({
    "Feature": feature_names,
    "Importance": importances
})

# Sort from highest to lowest
importance_df = importance_df.sort_values(
    by="Importance",
    ascending=False
)

# Display top 15 features
display(importance_df.head(15).round(4))
```

```python
top_features = importance_df.head(15).sort_values(
    by="Importance"
)

plt.figure(figsize=(10, 7))

plt.barh(
    top_features["Feature"],
    top_features["Importance"]
)

plt.xlabel("Feature Importance")
plt.title("Top 15 Features Affecting House Price Predictions")
plt.tight_layout()
plt.show()
```

```python
from sklearn.model_selection import RandomizedSearchCV

# Use a sample to make tuning faster
X_tune = X_train.sample(
    n=min(20000, len(X_train)),
    random_state=42
)

y_tune = y_train.loc[X_tune.index]

# Parameters to test
param_grid = {
    "regressor__n_estimators": [60, 100],
    "regressor__max_depth": [12, 20, None],
    "regressor__min_samples_leaf": [2, 5, 10],
    "regressor__max_features": [0.7, 1.0]
}

search = RandomizedSearchCV(
    estimator=rf_model,
    param_distributions=param_grid,
    n_iter=6,
    cv=3,
    scoring="neg_mean_absolute_error",
    random_state=42,
    n_jobs=-1,
    verbose=1
)

search.fit(X_tune, y_tune)

print("Best parameters:")
print(search.best_params_)

print("Best CV MAE:", round(-search.best_score_, 2), "lakhs")

```

```python
best_model = search.best_estimator_

# Retrain on all training rows
best_model.fit(X_train, y_train)

# Evaluate on the untouched test set
best_pred = best_model.predict(X_test)

best_mae = mean_absolute_error(y_test, best_pred)
best_rmse = np.sqrt(mean_squared_error(y_test, best_pred))
best_r2 = r2_score(y_test, best_pred)

print("Tuned Random Forest")
print("MAE:", round(best_mae, 2), "lakhs")
print("RMSE:", round(best_rmse, 2), "lakhs")
print("R2 Score:", round(best_r2, 4))
```

```python
from sklearn.compose import TransformedTargetRegressor
from sklearn.base import clone
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Clone the tuned model
base_model = clone(best_model)

# Train using log-transformed prices
log_model = TransformedTargetRegressor(
    regressor=base_model,
    func=np.log1p,
    inverse_func=np.expm1
)

log_model.fit(X_train, y_train)

# Predictions are automatically converted back to lakhs
log_pred = log_model.predict(X_test)

# Evaluate on the original price scale
log_mae = mean_absolute_error(y_test, log_pred)
log_rmse = np.sqrt(mean_squared_error(y_test, log_pred))
log_r2 = r2_score(y_test, log_pred)

print("Log-Transformed Random Forest")
print("MAE:", round(log_mae, 2), "lakhs")
print("RMSE:", round(log_rmse, 2), "lakhs")
print("R2 Score:", round(log_r2, 4))
```

```python

import gradio as gr
import pandas as pd
import matplotlib.pyplot as plt
import tempfile
import os

# =========================================================
# 1. CHECK MODEL VARIABLES
# =========================================================
required_variables = [
    "best_model", "X_test", "y_test", "best_pred",
    "best_mae", "best_rmse", "best_r2"
]

missing_variables = [
    name for name in required_variables
    if name not in globals()
]

if missing_variables:
    raise ValueError(
        "Please run the model training and evaluation cells first. "
        "Missing variables: " + ", ".join(missing_variables)
    )

# Session-only prediction history
prediction_history = []


# =========================================================
# 2. HOUSE PRICE PREDICTION
# =========================================================
def predict_house_price(
    city, locality, property_type, furnishing,
    bhk, bathrooms, super_area, carpet_area,
    floor_number, total_floors, age,
    parking, lift, gated, metro_distance, center_distance
):
    # Validate inputs
    if super_area is None or carpet_area is None:
        return (
            "<h3 style='color:#b91c1c'>Please enter both area values.</h3>",
            pd.DataFrame(prediction_history)
        )

    if super_area <= 0 or carpet_area <= 0:
        return (
            "<h3 style='color:#b91c1c'>Area must be greater than zero.</h3>",
            pd.DataFrame(prediction_history)
        )

    if carpet_area > super_area:
        return (
            "<h3 style='color:#b91c1c'>Carpet area cannot exceed super area.</h3>",
            pd.DataFrame(prediction_history)
        )

    if floor_number > total_floors:
        return (
            "<h3 style='color:#b91c1c'>Floor cannot exceed total floors.</h3>",
            pd.DataFrame(prediction_history)
        )

    # Build input with the exact training feature names
    input_data = pd.DataFrame([{
        "City": city,
        "Locality_Type": locality,
        "Property_Type": property_type,
        "BHK": int(bhk),
        "Bathrooms": int(bathrooms),
        "Super_Area_SqFt": float(super_area),
        "Carpet_Area_SqFt": float(carpet_area),
        "Floor_Number": int(floor_number),
        "Total_Floors": int(total_floors),
        "Age_of_Property": int(age),
        "Furnishing_Status": furnishing,
        "Parking": int(parking),
        "Lift_Available": int(lift),
        "Gated_Community": int(gated),
        "Distance_to_Metro_km": float(metro_distance),
        "Distance_to_City_Center_km": float(center_distance)
    }])

    # Predict using the trained model
    prediction = float(best_model.predict(input_data)[0])

    if prediction <= 0:
        return (
            "<h3 style='color:#b91c1c'>Unable to generate a positive estimate.</h3>",
            pd.DataFrame(prediction_history)
        )

    # Save successful prediction
    prediction_history.append({
        "City": city,
        "Locality": locality,
        "Property Type": property_type,
        "Furnishing": furnishing,
        "BHK": int(bhk),
        "Bathrooms": int(bathrooms),
        "Super Area (sq ft)": float(super_area),
        "Carpet Area (sq ft)": float(carpet_area),
        "Predicted Price (₹ lakh)": round(prediction, 2),
        "Predicted Price (₹ crore)": round(prediction / 100, 3)
    })

    # Clearly visible prediction card
    result_html = f"""
    <div style="
        padding: 28px;
        border-radius: 18px;
        background-color: #ecfdf5;
        border: 2px solid #10b981;
        color: #14532d;
        font-family: Arial, sans-serif;
    ">
        <h2 style="color: #065f46; margin-bottom: 16px;">
            🏠 Estimated House Price
        </h2>

        <div style="
            background-color: #ffffff;
            padding: 20px;
            border-radius: 12px;
            margin: 15px 0;
            border: 1px solid #a7f3d0;
        ">
            <p style="color: #047857; margin: 0; font-size: 16px;">
                Predicted Property Value
            </p>

            <h1 style="
                color: #065f46;
                font-size: 38px;
                margin: 10px 0;
            ">
                ₹{prediction:,.2f} Lakh
            </h1>

            <h2 style="color: #047857; margin: 0;">
                ₹{prediction / 100:,.3f} Crore
            </h2>
        </div>

        <p style="color: #14532d;">
            <b>City:</b> {city}
        </p>

        <p style="color: #14532d;">
            <b>Locality:</b> {locality}
        </p>

        <p style="color: #14532d;">
            <b>Property Type:</b> {property_type} · {bhk} BHK
        </p>

        <p style="color: #14532d;">
            <b>Super Area:</b> {super_area:,.0f} sq ft
        </p>

        <p style="color: #14532d;">
            <b>Furnishing:</b> {furnishing}
        </p>

        <p style="
            color: #475569;
            font-size: 13px;
            margin-top: 20px;
        ">
            Educational ML estimate only, not a professional valuation.
        </p>
    </div>
    """

    return result_html, pd.DataFrame(prediction_history)


# =========================================================
# 3. HISTORY FUNCTIONS
# =========================================================
def refresh_history():
    return pd.DataFrame(prediction_history)


def clear_history():
    prediction_history.clear()
    return pd.DataFrame(prediction_history)


def download_history():
    if not prediction_history:
        return None

    history_df = pd.DataFrame(prediction_history)

    with tempfile.NamedTemporaryFile(
        suffix=".csv",
        delete=False
    ) as temp_file:
        file_path = temp_file.name

    history_df.to_csv(file_path, index=False)
    return file_path


# =========================================================
# 4. ACTUAL VS PREDICTED GRAPH
# =========================================================
def create_prediction_plot():
    fig, ax = plt.subplots(figsize=(8, 5))

    actual = pd.Series(y_test).reset_index(drop=True)
    predicted = pd.Series(best_pred).reset_index(drop=True)

    ax.scatter(actual, predicted, alpha=0.3)

    min_value = min(float(actual.min()), float(predicted.min()))
    max_value = max(float(actual.max()), float(predicted.max()))

    ax.plot(
        [min_value, max_value],
        [min_value, max_value],
        linestyle="--"
    )

    ax.set_xlabel("Actual Price (₹ lakh)")
    ax.set_ylabel("Predicted Price (₹ lakh)")
    ax.set_title("Actual vs Predicted House Prices")
    fig.tight_layout()

    return fig


# =========================================================
# 5. BUILD THE APP
# =========================================================
with gr.Blocks(
    title="House Price Prediction",
    theme=gr.themes.Soft()
) as demo:

    gr.Markdown("# 🏠 House Price Prediction")
    gr.Markdown(
        "Estimate a property's price using our trained Random Forest model."
    )

    # Location and property
    with gr.Group():
        gr.Markdown("## 📍 Location and Property")

        with gr.Row():
            city = gr.Dropdown(
                choices=[
                    "Ahmedabad", "Bangalore", "Chandigarh",
                    "Chennai", "Delhi", "Hyderabad",
                    "Jaipur", "Kolkata", "Mumbai", "Pune"
                ],
                value="Mumbai",
                label="City"
            )

            locality = gr.Dropdown(
                ["Affordable", "Mid-Range", "Premium"],
                value="Premium",
                label="Locality Type"
            )

        with gr.Row():
            property_type = gr.Dropdown(
                ["Apartment", "Villa", "Builder Floor", "Independent House"],
                value="Apartment",
                label="Property Type"
            )

            furnishing = gr.Dropdown(
                ["Furnished", "Semi-Furnished", "Unfurnished"],
                value="Furnished",
                label="Furnishing Status"
            )

    # Size and layout
    with gr.Group():
        gr.Markdown("## 📐 Size and Layout")

        with gr.Row():
            bhk = gr.Slider(1, 6, value=3, step=1, label="BHK")
            bathrooms = gr.Slider(1, 6, value=3, step=1, label="Bathrooms")

        with gr.Row():
            super_area = gr.Number(
                value=1500,
                label="Super Area (sq ft)"
            )
            carpet_area = gr.Number(
                value=1100,
                label="Carpet Area (sq ft)"
            )

        with gr.Row():
            floor_number = gr.Slider(
                0, 43, value=5, step=1, label="Floor Number"
            )
            total_floors = gr.Slider(
                1, 44, value=20, step=1, label="Total Floors"
            )
            age = gr.Slider(
                0, 39, value=5, step=1, label="Property Age (years)"
            )

    # Amenities and distances
    with gr.Group():
        gr.Markdown("## 🛠️ Amenities and Distances")

        with gr.Row():
            parking = gr.Radio(
                [0, 1], value=1, label="Parking Available"
            )
            lift = gr.Radio(
                [0, 1], value=1, label="Lift Available"
            )
            gated = gr.Radio(
                [0, 1], value=1, label="Gated Community"
            )

        with gr.Row():
            metro_distance = gr.Number(
                value=1.5,
                label="Distance to Metro (km)"
            )
            center_distance = gr.Number(
                value=5.0,
                label="Distance to City Center (km)"
            )

    predict_button = gr.Button(
        "🔮 Predict House Price",
        variant="primary"
    )

    result_output = gr.HTML(label="Prediction Result")

    # Prediction history
    gr.Markdown("---")
    gr.Markdown("## 📋 Prediction History")

    history_table = gr.Dataframe(
        value=pd.DataFrame(prediction_history),
        interactive=False,
        label="Previous Predictions"
    )

    with gr.Row():
        refresh_button = gr.Button("🔄 Refresh History")
        clear_button = gr.Button("🗑️ Clear History")
        download_button = gr.Button("📥 Prepare CSV Download")

    download_file = gr.File(label="Download Prediction History")

    predict_button.click(
        fn=predict_house_price,
        inputs=[
            city, locality, property_type, furnishing,
            bhk, bathrooms, super_area, carpet_area,
            floor_number, total_floors, age,
            parking, lift, gated,
            metro_distance, center_distance
        ],
        outputs=[result_output, history_table]
    )

    refresh_button.click(
        fn=refresh_history,
        inputs=[],
        outputs=history_table
    )

    clear_button.click(
        fn=clear_history,
        inputs=[],
        outputs=history_table
    )

    download_button.click(
        fn=download_history,
        inputs=[],
        outputs=download_file
    )

    # Performance dashboard
    gr.Markdown("---")
    gr.Markdown("# 📈 Model Performance Dashboard")

    with gr.Row():
        gr.Number(
            value=round(float(best_mae), 2),
            label="Test MAE (₹ lakh)",
            interactive=False
        )
        gr.Number(
            value=round(float(best_rmse), 2),
            label="Test RMSE (₹ lakh)",
            interactive=False
        )
        gr.Number(
            value=round(float(best_r2), 4),
            label="Test R² Score",
            interactive=False
        )

    gr.Markdown(
        "These metrics describe performance on the held-out test set; "
        "they do not guarantee the accuracy of an individual prediction."
    )

    gr.Plot(
        value=create_prediction_plot(),
        label="Actual vs Predicted Prices"
    )

    gr.Markdown(
        "**Disclaimer:** This model uses a synthetic dataset. "
        "Results are educational estimates, not verified market valuations."
    )


# =========================================================
# 6. LAUNCH
# =========================================================
demo.launch(share=True)
```

```python

import gradio as gr
import pandas as pd
import matplotlib.pyplot as plt
import tempfile
import os

# =========================================================
# 1. CHECK MODEL VARIABLES
# =========================================================
required_variables = [
    "best_model", "X_test", "y_test", "best_pred",
    "best_mae", "best_rmse", "best_r2"
]

missing_variables = [
    name for name in required_variables
    if name not in globals()
]

if missing_variables:
    raise ValueError(
        "Please run the model training and evaluation cells first. "
        "Missing variables: " + ", ".join(missing_variables)
    )

# Session-only prediction history
prediction_history = []


# =========================================================
# 2. HOUSE PRICE PREDICTION
# =========================================================
def predict_house_price(
    city, locality, property_type, furnishing,
    bhk, bathrooms, super_area, carpet_area,
    floor_number, total_floors, age,
    parking, lift, gated, metro_distance, center_distance
):
    # Validate inputs
    if super_area is None or carpet_area is None:
        return (
            "<h3 style='color:#b91c1c'>Please enter both area values.</h3>",
            pd.DataFrame(prediction_history)
        )

    if super_area <= 0 or carpet_area <= 0:
        return (
            "<h3 style='color:#b91c1c'>Area must be greater than zero.</h3>",
            pd.DataFrame(prediction_history)
        )

    if carpet_area > super_area:
        return (
            "<h3 style='color:#b91c1c'>Carpet area cannot exceed super area.</h3>",
            pd.DataFrame(prediction_history)
        )

    if floor_number > total_floors:
        return (
            "<h3 style='color:#b91c1c'>Floor cannot exceed total floors.</h3>",
            pd.DataFrame(prediction_history)
        )

    # Build input with the exact training feature names
    input_data = pd.DataFrame([{
        "City": city,
        "Locality_Type": locality,
        "Property_Type": property_type,
        "BHK": int(bhk),
        "Bathrooms": int(bathrooms),
        "Super_Area_SqFt": float(super_area),
        "Carpet_Area_SqFt": float(carpet_area),
        "Floor_Number": int(floor_number),
        "Total_Floors": int(total_floors),
        "Age_of_Property": int(age),
        "Furnishing_Status": furnishing,
        "Parking": int(parking),
        "Lift_Available": int(lift),
        "Gated_Community": int(gated),
        "Distance_to_Metro_km": float(metro_distance),
        "Distance_to_City_Center_km": float(center_distance)
    }])

    # Predict using the trained model
    prediction = float(best_model.predict(input_data)[0])

    if prediction <= 0:
        return (
            "<h3 style='color:#b91c1c'>Unable to generate a positive estimate.</h3>",
            pd.DataFrame(prediction_history)
        )

    # Save successful prediction
    prediction_history.append({
        "City": city,
        "Locality": locality,
        "Property Type": property_type,
        "Furnishing": furnishing,
        "BHK": int(bhk),
        "Bathrooms": int(bathrooms),
        "Super Area (sq ft)": float(super_area),
        "Carpet Area (sq ft)": float(carpet_area),
        "Predicted Price (₹ lakh)": round(prediction, 2),
        "Predicted Price (₹ crore)": round(prediction / 100, 3)
    })

    # Clearly visible prediction card
    result_html = f"""
    <div style="
        padding: 28px;
        border-radius: 18px;
        background-color: #ecfdf5;
        border: 2px solid #10b981;
        color: #14532d;
        font-family: Arial, sans-serif;
    ">
        <h2 style="color: #065f46; margin-bottom: 16px;">
            🏠 Estimated House Price
        </h2>

        <div style="
            background-color: #ffffff;
            padding: 20px;
            border-radius: 12px;
            margin: 15px 0;
            border: 1px solid #a7f3d0;
        ">
            <p style="color: #047857; margin: 0; font-size: 16px;">
                Predicted Property Value
            </p>

            <h1 style="
                color: #065f46;
                font-size: 38px;
                margin: 10px 0;
            ">
                ₹{prediction:,.2f} Lakh
            </h1>

            <h2 style="color: #047857; margin: 0;">
                ₹{prediction / 100:,.3f} Crore
            </h2>
        </div>

        <p style="color: #14532d;">
            <b>City:</b> {city}
        </p>

        <p style="color: #14532d;">
            <b>Locality:</b> {locality}
        </p>

        <p style="color: #14532d;">
            <b>Property Type:</b> {property_type} · {bhk} BHK
        </p>

        <p style="color: #14532d;">
            <b>Super Area:</b> {super_area:,.0f} sq ft
        </p>

        <p style="color: #14532d;">
            <b>Furnishing:</b> {furnishing}
        </p>

        <p style="
            color: #475569;
            font-size: 13px;
            margin-top: 20px;
        ">
            Educational ML estimate only, not a professional valuation.
        </p>
    </div>
    """

    return result_html, pd.DataFrame(prediction_history)


# =========================================================
# 3. HISTORY FUNCTIONS
# =========================================================
def refresh_history():
    return pd.DataFrame(prediction_history)


def clear_history():
    prediction_history.clear()
    return pd.DataFrame(prediction_history)


def download_history():
    if not prediction_history:
        return None

    history_df = pd.DataFrame(prediction_history)

    with tempfile.NamedTemporaryFile(
        suffix=".csv",
        delete=False
    ) as temp_file:
        file_path = temp_file.name

    history_df.to_csv(file_path, index=False)
    return file_path


# =========================================================
# 4. ACTUAL VS PREDICTED GRAPH
# =========================================================
def create_prediction_plot():
    fig, ax = plt.subplots(figsize=(8, 5))

    actual = pd.Series(y_test).reset_index(drop=True)
    predicted = pd.Series(best_pred).reset_index(drop=True)

    ax.scatter(actual, predicted, alpha=0.3)

    min_value = min(float(actual.min()), float(predicted.min()))
    max_value = max(float(actual.max()), float(predicted.max()))

    ax.plot(
        [min_value, max_value],
        [min_value, max_value],
        linestyle="--"
    )

    ax.set_xlabel("Actual Price (₹ lakh)")
    ax.set_ylabel("Predicted Price (₹ lakh)")
    ax.set_title("Actual vs Predicted House Prices")
    fig.tight_layout()

    return fig


# =========================================================
# 5. BUILD THE APP
# =========================================================
with gr.Blocks(
    title="House Price Prediction",
    theme=gr.themes.Soft()
) as demo:

    gr.Markdown("# 🏠 House Price Prediction")
    gr.Markdown(
        "Estimate a property's price using our trained Random Forest model."
    )

    # Location and property
    with gr.Group():
        gr.Markdown("## 📍 Location and Property")

        with gr.Row():
            city = gr.Dropdown(
                choices=[
                    "Ahmedabad", "Bangalore", "Chandigarh",
                    "Chennai", "Delhi", "Hyderabad",
                    "Jaipur", "Kolkata", "Mumbai", "Pune"
                ],
                value="Mumbai",
                label="City"
            )

            locality = gr.Dropdown(
                ["Affordable", "Mid-Range", "Premium"],
                value="Premium",
                label="Locality Type"
            )

        with gr.Row():
            property_type = gr.Dropdown(
                ["Apartment", "Villa", "Builder Floor", "Independent House"],
                value="Apartment",
                label="Property Type"
            )

            furnishing = gr.Dropdown(
                ["Furnished", "Semi-Furnished", "Unfurnished"],
                value="Furnished",
                label="Furnishing Status"
            )

    # Size and layout
    with gr.Group():
        gr.Markdown("## 📐 Size and Layout")

        with gr.Row():
            bhk = gr.Slider(1, 6, value=3, step=1, label="BHK")
            bathrooms = gr.Slider(1, 6, value=3, step=1, label="Bathrooms")

        with gr.Row():
            super_area = gr.Number(
                value=1500,
                label="Super Area (sq ft)"
            )
            carpet_area = gr.Number(
                value=1100,
                label="Carpet Area (sq ft)"
            )

        with gr.Row():
            floor_number = gr.Slider(
                0, 43, value=5, step=1, label="Floor Number"
            )
            total_floors = gr.Slider(
                1, 44, value=20, step=1, label="Total Floors"
            )
            age = gr.Slider(
                0, 39, value=5, step=1, label="Property Age (years)"
            )

    # Amenities and distances
    with gr.Group():
        gr.Markdown("## 🛠️ Amenities and Distances")

        with gr.Row():
            parking = gr.Radio(
                [0, 1], value=1, label="Parking Available"
            )
            lift = gr.Radio(
                [0, 1], value=1, label="Lift Available"
            )
            gated = gr.Radio(
                [0, 1], value=1, label="Gated Community"
            )

        with gr.Row():
            metro_distance = gr.Number(
                value=1.5,
                label="Distance to Metro (km)"
            )
            center_distance = gr.Number(
                value=5.0,
                label="Distance to City Center (km)"
            )

    predict_button = gr.Button(
        "🔮 Predict House Price",
        variant="primary"
    )

    result_output = gr.HTML(label="Prediction Result")

    # Prediction history
    gr.Markdown("---")
    gr.Markdown("## 📋 Prediction History")

    history_table = gr.Dataframe(
        value=pd.DataFrame(prediction_history),
        interactive=False,
        label="Previous Predictions"
    )

    with gr.Row():
        refresh_button = gr.Button("🔄 Refresh History")
        clear_button = gr.Button("🗑️ Clear History")
        download_button = gr.Button("📥 Prepare CSV Download")

    download_file = gr.File(label="Download Prediction History")

    predict_button.click(
        fn=predict_house_price,
        inputs=[
            city, locality, property_type, furnishing,
            bhk, bathrooms, super_area, carpet_area,
            floor_number, total_floors, age,
            parking, lift, gated,
            metro_distance, center_distance
        ],
        outputs=[result_output, history_table]
    )

    refresh_button.click(
        fn=refresh_history,
        inputs=[],
        outputs=history_table
    )

    clear_button.click(
        fn=clear_history,
        inputs=[],
        outputs=history_table
    )

    download_button.click(
        fn=download_history,
        inputs=[],
        outputs=download_file
    )

    # Performance dashboard
    gr.Markdown("---")
    gr.Markdown("# 📈 Model Performance Dashboard")

    with gr.Row():
        gr.Number(
            value=round(float(best_mae), 2),
            label="Test MAE (₹ lakh)",
            interactive=False
        )
        gr.Number(
            value=round(float(best_rmse), 2),
            label="Test RMSE (₹ lakh)",
            interactive=False
        )
        gr.Number(
            value=round(float(best_r2), 4),
            label="Test R² Score",
            interactive=False
        )

    gr.Markdown(
        "These metrics describe performance on the held-out test set; "
        "they do not guarantee the accuracy of an individual prediction."
    )

    gr.Plot(
        value=create_prediction_plot(),
        label="Actual vs Predicted Prices"
    )

    gr.Markdown(
        "**Disclaimer:** This model uses a synthetic dataset. "
        "Results are educational estimates, not verified market valuations."
    )


# =========================================================
# 6. LAUNCH
# =========================================================
demo.launch(share=True)
```
